Task 2: Create Passenger Booking Function

```python
def book_ticket(passenger_name,train_name):
    print("Passenger Name : ", passenger_name)
    print("Train Name : ", train_name)  
    print("Reservation:Confirmed")

book_ticket("Ram","Rajdhani Express")
```

```python
def calculate_fare(ticket_price,number_of_passengers):
    total_fare = ticket_price * number_of_passengers
    print("Total Fare: ", total_fare)   
calculate_fare(500, 3)
```
