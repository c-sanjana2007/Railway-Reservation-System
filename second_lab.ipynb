Creating Variables for Railway Information


```python

train_name = "Rajdhani Express"
train_number = 12952
source_station="Mumbai"
destination_station="Delhi"
ticket_price = 2450.75
print("Welcome to Indian Railway Reservation System")
print("Train Name: ",train_name)
print("Train Number: ",train_number)
print("Source Station: ",source_station)
print("Destination Station: ",destination_station)
```

```python
# TYPE FUNCTION
print(type(train_name))
print(type(train_number))
print(type(ticket_price))
```

Task : Identifiers & Literals


```python
passenger_name="Rahul Sharma"
seat_number="B2-45"
is_confirmed=True
print("-----------Railway Ticket----------------")
print("Passenger Name : ",passenger_name)
print("Seat Number : ",seat_number)
print("Ticket Status : ",is_confirmed)
print("------------------------------------------")
```

```python
#Typecasting

ticket_price_integer = int(ticket_price)
print(type(ticket_price_integer))         #  int type

train_number_string = str(train_number)   #  Convert int-->str
print(type(train_number_string))          # str type
```
