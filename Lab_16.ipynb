```python
class Car:
    def __init__(self):
        print("Ye class ka constructor hai")
    def display(self):
        print("Your car is awesome")
        
class AutomaticCar(Car):
    def __init__(self):
        print("Ye automatic car ka constructor hai")
        super().__init__()
    def display(self):
        print("Your automatic car is awesome")
        super().display() 
auto = AutomaticCar()
auto.display()        
```

```python
class Parent:
    def parent_property(self):
        print('parent''s property')
class Brother(Parent):
    def brother_property(self):
        print('Child''s property')

class Sister(Parent):
    def sister_property(self):
        print('Grandchild''s property')
        
b = Brother()
b.brother_property()
b.parent_property()

s=Sister()
s.sister_property()
s.parent_property()

p=Parent()
```
