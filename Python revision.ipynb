TASK 1: append() & insert()

```python
subjects=["Python","Tableau","Excel","Power BI","ML"]
print(subjects)

#Add one more subject at the end of the list

subjects.append("Statistics")
print(subjects)

#Add a subject at the 2nd position(index 1) in the list

subjects.insert(1,"SQL")
print(subjects)
```

Task 2: remove(), pop() and clear()

```python
#Create a list named numbers with 6 number of your choice.
numbers=[1,2,3,4,5,6]

#Remove one number from the list by its value
numbers.remove(2)                    

#Remove the last number from the list using a method that also returns the removed value
print(numbers.pop())                   # 6

#Clear
#clear can clear the whole list means if we use numbers.clear() then from the the list whole value will removed and list will empty
# output []
```

Task 3 : sort(), reverse(), and count()

```python
#Create a list named marks with 6 numbers, with atleast one number repeated
marks=[50,80,75,40,60,75]

#Sort the list in ascending order
marks.sort()

#Reverse the sorted list
marks.reverse()

#Count how many times the repeated  number appears in the list.
marks.count(75)      
```
